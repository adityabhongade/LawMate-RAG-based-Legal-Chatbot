```python
import json
import glob

json_files = ["/kaggle/input/cleaned-lawmate/cleaned_constitution_qa.json", "/kaggle/input/cleaned-lawmate/cleaned_crpc_qa.json", "/kaggle/input/cleaned-lawmate/cleaned_ipc_qa.json"]

combined_data = []

for file in json_files:
    with open(file, "r", encoding="utf-8") as f:
        data = json.load(f)
        combined_data.extend(data)

with open("combined_knowledge_base.json", "w", encoding="utf-8") as f:
    json.dump(combined_data, f, indent=4, ensure_ascii=False)

print("Merged JSON saved as combined_knowledge_base.json")
```
