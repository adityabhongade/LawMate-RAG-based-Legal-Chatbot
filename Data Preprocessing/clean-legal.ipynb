```python
import os
import json
import pandas as pd
```

```python
const_path = 'constitution_qa.json'
crpc_path = 'crpc_qa.json'
ipc_path = 'ipc_qa.json'

files = [const_path, crpc_path, ipc_path]

result_dir = 'Cleaned_Data'
os.makedirs(result_dir, exist_ok=True)
```

```python
for file in files:
    file = '/kaggle/input/llm-fine-tuning-dataset-of-indian-legal-texts/' + file
    with open(file, 'r', encoding='utf-8') as f:
        data = json.load(f)

    cleaned_data = []
    for item in data:
        if 'question' in item and 'answer' in item:
            question = item['question'].strip()
            answer = item['answer'].strip()
            cleaned_data.append({'question': question, 'answer': answer})

    df = pd.DataFrame(cleaned_data)
    df = df.drop_duplicates()

    filename = os.path.basename(file)
    output_file = os.path.join(result_dir, f'cleaned_{filename}')
    df.to_json(output_file, orient='records', indent=4, force_ascii=False)
```

```python
import shutil

shutil.make_archive(result_dir, 'zip', result_dir)
```
